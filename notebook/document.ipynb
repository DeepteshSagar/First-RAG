```python
## Create a simple txt file

import os
os.makedirs("../data/text_files", exist_ok=True)
```

```python
about_rag={
    "../data/text_files/about_rag.txt":"""RAG (Retrieval-Augmented Generation)

RAG stands for Retrieval-Augmented Generation. It is a technique used with Large Language Models (LLMs) to make their responses more accurate, relevant, and based on specific external information.

Normally, an LLM answers using the knowledge it learned during training. With RAG, the model can search an external knowledge source, such as PDFs, documents, databases, or websites, and use the retrieved information to answer the user's question.

How RAG works
User Query - The user asks a question.
Retrieval - The system searches a knowledge base for relevant information.
Augmentation - The retrieved information is added to the user's query as context.
Generation - The LLM uses this context to generate the final answer.

Example:
Suppose a college has a PDF containing its rules and regulations. If a student asks, "What is the attendance requirement?", a RAG system searches the PDF, retrieves the relevant section, and provides it to the LLM. The LLM then generates an answer based on that section.

Advantages of RAG
Reduces incorrect or hallucinated answers.
Can work with private and custom data.
Information can be updated without retraining the LLM.
Useful for document-based question answering and chatbots.

In simple terms:

RAG gives an LLM the ability to look up relevant information before answering a question."""
}

for file_path, content in about_rag.items():
    with open(file_path, 'w', encoding="utf-8") as f:
        f.write(content)

print("Sample text file created!")
```

```python
###Textloader
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/about_rag.txt", encoding = "utf-8")
document = loader.load()
print(f"Loaded {len(document)} document(s) from the text file.")
print(document)
```

```python
### Directory Loader
from langchain_community.document_loaders import DirectoryLoader

## Load all the text files from the directory
loader = DirectoryLoader(
    "../data/text_files", 
     glob="*.txt", ##Pattern to match files
     loader_cls = TextLoader, ##loader class to use
     show_progress=False, ##Show progress bar
     loader_kwargs = {"encoding": "utf-8"}, ##Encoding of the files
     recursive=True ##Load files from subdirectories as well

)

documents=loader.load()
print(documents)
```

```python
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

## Load all the text files from the directory
loader = DirectoryLoader(
    "../data/pdf", 
     glob="*.pdf", ##Pattern to match files
     loader_cls = PyMuPDFLoader, ##loader class to use
     show_progress=False, ##Show progress bar
     recursive=True ##Load files from subdirectories as well

)

pdf_documents=loader.load()
pdf_documents
```
